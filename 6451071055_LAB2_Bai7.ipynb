### Chuẩn bị các thư viện cần thiết

```python
%matplotlib inline
import matplotlib
import matplotlib.pyplot as plt

from sklearn import datasets, tree, metrics
from sklearn.model_selection import train_test_split
```

### Download dữ liệu có sẵn trong thư viện sklearn

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# The digits dataset
digits = datasets.load_digits()
print('Số mẫu:', len(digits.images), '- Kích thước mỗi ảnh:', digits.images[0].shape)
```

### Biểu diễn 4 mẫu dữ liệu huấn luyện tương ứng với 4 lớp 0, 1, 2, 3

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
_, axes = plt.subplots(1, 4)
images_and_labels = list(zip(digits.images, digits.target))
for ax, (image, label) in zip(axes, images_and_labels[:4]):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Training: %i' % label)
plt.show()
```

### Chuyển mỗi ảnh từ ma trận 8x8 thành ma trận 1x64

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# To apply a classifier on this data, we need to flatten the image, to
# turn the data in a (samples, feature) matrix:
n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))
print(data.shape)
```

### Sử dụng thuật toán cây ID3

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# Create a classifier: a decision tree classifier
classifier = tree.DecisionTreeClassifier(criterion="entropy", random_state=33)
```

### Chia dữ liệu thành hai phần: huấn luyện 80%, kiểm thử 20%

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# Split data into train and test subsets
X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size=0.2, shuffle=False)
print('Huấn luyện:', len(X_train), '- Kiểm thử:', len(X_test))
```

### Tiến hành xây dựng cây ID3

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# We learn the digits on the first part of the digits
classifier.fit(X_train, y_train)
```

### Dự đoán nhãn dữ liệu cho tập kiểm thử

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# Now predict the value of the digit on the second part:
predicted = classifier.predict(X_test)
print(predicted[:10])
```

### Biểu diễn một vài kết quả dự đoán

Tài liệu dùng `digits.images[n_samples // 2:]`, là cách chia 50/50 nên ảnh không khớp với nhãn dự đoán khi chia 80/20. Ở đây lấy đúng các ảnh của tập kiểm thử bằng `digits.images[-len(X_test):]`.

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
_, axes = plt.subplots(1, 4)
images_and_predictions = list(zip(digits.images[-len(X_test):], predicted))
for ax, (image, prediction) in zip(axes, images_and_predictions[:4]):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Prediction: %i' % prediction)
plt.show()
```

### Thông số đạt được của cây ID3

`metrics.plot_confusion_matrix` trong tài liệu đã bị xóa từ scikit-learn 1.2 nên được thay bằng `metrics.ConfusionMatrixDisplay.from_estimator`.

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
print("Classification report for classifier %s:\n%s\n"
      % (classifier, metrics.classification_report(y_test, predicted)))
disp = metrics.ConfusionMatrixDisplay.from_estimator(classifier, X_test, y_test)
disp.figure_.suptitle("Confusion Matrix")
print("Confusion matrix:\n%s" % disp.confusion_matrix)
plt.show()
```

### Cài đặt thư viện xử lý hình ảnh pillow và import vào

```python
from PIL import Image, ImageOps
import numpy as np
```

### Nhận dạng ảnh chữ số viết tay thứ nhất (test_image_Bai7(1).jpg)

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# Open image
img = Image.open('data/test_image_Bai7(1).jpg').convert("L").resize((8, 8))
# Invert image
img = ImageOps.invert(img)
# Convert to numpy array
im2arr = np.array(img)
# Show
plt.imshow(im2arr, cmap=plt.cm.gray_r, interpolation='nearest')
plt.title('Ảnh gốc: chữ số 1')
plt.show()
```

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# Change from 2d array (8x8) to 1d array (1x64)
img1d = im2arr.reshape([1, 64])
# Normalize black & white image array (ngưỡng 110)
img1d[img1d > 109] = 155
img1d[img1d < 110] = 0
print(img1d)

# After normalize
plt.imshow(im2arr, cmap=plt.cm.gray_r, interpolation='nearest')
plt.show()
```

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
# Predict label of number
y_pred = classifier.predict(img1d)
print('Chữ số thực tế: 1 - Mô hình dự đoán:', y_pred)
```

### Kiểm tra độ nhạy của kết quả với ngưỡng phân biệt đen, trắng

```python
print("Đinh Nhật Huyền Nhân - 6451071055")
for path, truth in [('data/test_image_Bai7(1).jpg', 1), ('data/test_image_Bai7.jpg', 2)]:
    gray = np.array(ImageOps.invert(Image.open(path).convert("L").resize((8, 8))))
    for threshold in [90, 100, 110, 120]:
        binary = (gray >= threshold).astype('uint8') * 155
        print(f'Chữ số {truth} - ngưỡng {threshold}: dự đoán {classifier.predict(binary.reshape([1, 64]))}')
```

**Nhận xét:** Cây ID3 đạt accuracy 0,80 trên 360 mẫu kiểm thử, ma trận nhầm lẫn khớp với tài liệu. Với ngưỡng 110, mô hình nhận đúng cả hai ảnh (chữ số 1 và 2). Kết quả phụ thuộc vào ngưỡng: ảnh chữ số 2 bị nhận thành 5 khi dùng ngưỡng 100, nên cần chọn ngưỡng phù hợp với điều kiện ảnh chụp. Giá trị 155 nằm ngoài thang 0-16 của dữ liệu digits, nhưng cây chỉ so sánh với các ngưỡng trong thang đó nên 155 có tác dụng như một điểm ảnh đậm tối đa. Hai ảnh là quá ít để đánh giá khả năng nhận dạng ảnh thật.
